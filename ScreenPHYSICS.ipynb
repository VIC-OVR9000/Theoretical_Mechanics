# The Near-Field Deviation Index (NDI)
### *Quantifying the Failure of the Inverse Square Law in Modern Display Ergonomics*

---

## 1. Introduction: The Point Source Fallacy
In introductory physics, light intensity ($I$) is typically governed by the **Inverse Square Law**:

$$I_{point}(r) = \frac{L}{r^2}$$

This model assumes a **point source**—a single, infinitely small origin. While this works for distant stars or lightbulbs, it fails significantly when applied to **extended sources** like computer monitors, tablets, and televisions at typical viewing distances.

### The Problem
Modern displays are large, flat, **Lambertian emitters**. When an observer is in the "Near-Field" (e.g., sitting at a desk), the light reaching the eye is an aggregate of millions of pixels spread across a 2D plane. The differential light contribution from an area element $dA$ is defined as:

$$dI = \frac{L \cdot \cos(\theta) \cdot dA}{s^2}$$

Because the light from the edges of the screen arrives at different angles and distances than the light from the center, the total intensity does **not** drop off by the square of the distance. To find the true intensity, we must evaluate the double integral across the screen boundaries $(w, h)$:

$$I(r) = \int_{-h/2}^{h/2} \int_{-w/2}^{w/2} \frac{L \cdot r}{(x^2 + y^2 + r^2)^{3/2}} \, dx \, dy$$

---

## 2. Project Objective
This notebook derives an accurate closed-form model for rectangular emitters and introduces the **Near-Field Deviation Index (NDI)**.

**Key Mathematical Models:**

1. **The Closed-Form Analytical Solution ($I_{rect}$):**
   $$I_{rect}(r) = 4L \cdot \arctan\left(\frac{w \cdot h}{4r \sqrt{r^2 + (\frac{w}{2})^2 + (\frac{h}{2})^2}}\right)$$

2. **The Taylor Series Convergence (Far-Field):**
   Showing that as $r \to \infty$, and using $\arctan(x) \approx x$:
   $$I_{rect}(r) \approx 4L \cdot \frac{w \cdot h}{4r \sqrt{r^2}} \approx \frac{L \cdot (w \cdot h)}{r^2}$$

3. **The NDI Metric:**
   Quantifying the "Intensity Bulge" by calculating the area of deviation between the models:
   $$NDI = \int_{r_{min}}^{r_{max}} \left| I_{rect}(r) - I_{point}(r) \right| \, dr$$

4. **Average Intensity Concentration:**
   Measuring light "dilution" across the physical surface area $A$:
   $$I_{avg}(r) = \frac{I_{rect}(r)}{w \cdot h}$$

---

## 3. Variables & Units
* $r$: Radial distance from the screen center (meters).
* $w, h$: Physical width and height of the display (meters).
* $L$: Luminosity/Exitance of the panel (nits or $cd/m^2$).
* $NDI$: Dimensionless index of geometric deviation.


### **Mathematical Derivation of $I_{rect}$**

To derive the intensity of a rectangular screen, we treat the display as a **Lambertian surface** where every infinitesimal area element $dA$ acts as a point source following the Inverse Square Law.

---

**1. The Differential Element**
Consider a screen of width $w$ and height $h$ centered at the origin $(0,0,0)$ in the $xy$-plane. An observer is positioned at point $P(0,0,r)$ on the $z$-axis. An infinitesimal area element $dA = dx\,dy$ is located at $(x,y,0)$. The distance $s$ from the element to the observer is:

$$s = \sqrt{x^2 + y^2 + r^2}$$

**2. Applying Lambert's Cosine Law**
The light contribution $dI$ is proportional to the cosine of the angle $\theta$ between the surface normal (the $z$-axis) and the vector to the observer:

$$\cos(\theta) = \frac{r}{s} = \frac{r}{\sqrt{x^2 + y^2 + r^2}}$$

Following the radiometric definition for a differential area:

$$dI = \frac{L \cdot \cos(\theta) \cdot dA}{s^2} = \frac{L \cdot r}{(x^2 + y^2 + r^2)^{3/2}} dx\,dy$$

**3. The Double Integral Setup**
To find the total intensity $I(r)$, we integrate over the rectangular boundaries:

$$I(r) = \int_{-h/2}^{h/2} \int_{-w/2}^{w/2} \frac{L \cdot r}{(x^2 + y^2 + r^2)^{3/2}} \, dx \, dy$$

**4. First Integration ($x$-axis)**
Using the standard integral form $\int \frac{a}{(x^2 + C)^{3/2}} dx = \frac{ax}{C\sqrt{x^2+C}}$, we evaluate the inner integral:

$$\int_{-w/2}^{w/2} \frac{L \cdot r}{(x^2 + (y^2 + r^2))^{3/2}} dx = \left[ \frac{L \cdot r \cdot x}{(y^2 + r^2)\sqrt{x^2 + y^2 + r^2}} \right]_{-w/2}^{w/2}$$

Evaluating the limits:

$$= \frac{L \cdot r \cdot w}{(y^2 + r^2)\sqrt{(\frac{w}{2})^2 + y^2 + r^2}}$$

**5. Second Integration ($y$-axis)**
Integrating the result with respect to $y$ follows the form $\int \frac{1}{(y^2 + a^2)\sqrt{y^2 + b^2}} dy$, which yields an $\arctan$ solution:

$$I(r) = L \cdot w \cdot r \int_{-h/2}^{h/2} \frac{1}{(y^2 + r^2)\sqrt{y^2 + r^2 + (\frac{w}{2})^2}} dy$$

**6. Final Closed-Form Solution**
After simplifying the coefficients and evaluating the definite integral, we arrive at the analytical model for a rectangular emitter:

$$I_{rect}(r) = 4L \cdot \arctan\left(\frac{w \cdot h}{4r \sqrt{r^2 + (\frac{w}{2})^2 + (\frac{h}{2})^2}}\right)$$


### **Convergence Proof: The Far-Field Approximation**

To verify the validity of our analytical model $I_{rect}(r)$, we examine its behavior as the distance $r$ becomes much larger than the screen dimensions $w$ and $h$ ($r \to \infty$). At this limit, the screen should mathematically "collapse" into a single point source, returning us to the **Inverse Square Law**.

---

**1. Defining the Limit**
We start with our closed-form solution:
$$I_{rect}(r) = 4L \cdot \arctan\left(\frac{w \cdot h}{4r \sqrt{r^2 + (\frac{w}{2})^2 + (\frac{h}{2})^2}}\right)$$

As $r \to \infty$, the term inside the square root is dominated by $r^2$:
$$\sqrt{r^2 + (\frac{w}{2})^2 + (\frac{h}{2})^2} \approx \sqrt{r^2} = r$$

**2. Simplifying the Argument**
Substituting this approximation back into the $\arctan$ argument:
$$\text{Argument} \approx \frac{w \cdot h}{4r(r)} = \frac{w \cdot h}{4r^2}$$

**3. Applying the Taylor Expansion**
The Taylor Series expansion for $\arctan(x)$ when $x$ is very small is:
$$\arctan(x) = x - \frac{x^3}{3} + \frac{x^5}{5} - \dots$$
For large $r$, our argument $\frac{w \cdot h}{4r^2}$ approaches zero. Keeping only the first-order term:
$$\arctan\left(\frac{w \cdot h}{4r^2}\right) \approx \frac{w \cdot h}{4r^2}$$

**4. Final Reduction**
Multiplying by the leading coefficient $4L$:
$$I_{rect}(r) \approx 4L \cdot \left( \frac{w \cdot h}{4r^2} \right) = \frac{L \cdot (w \cdot h)}{r^2}$$

**5. Conclusion**
Since $(w \cdot h)$ represents the total area $A$ of the screen, the final form is:
$$I_{rect}(r) \approx \frac{L \cdot A}{r^2} \propto \frac{1}{r^2}$$

This proves that in the **Far-Field**, our model converges perfectly with the **Inverse Square Law**. The "Near-Field" effects we observe at close range are the result of the higher-order terms in the Taylor expansion that become significant as $r$ decreases.


### **Analysis: Behavior of Large-Area Emitters ($A > 1 \text{ m}^2$)**

When the screen area $A = w \cdot h$ exceeds $1 \text{ m}^2$ (e.g., large-format televisions or LED walls), the display enters the **Macro-Surface Regime**. In this state, the geometric deviation from the Inverse Square Law is at its maximum.

#### **1. Intensity Saturation**
As $A$ increases relative to $r$, the observer is "engulfed" by the Lambertian field. The $\arctan$ argument in our model approaches infinity, leading to a saturation of intensity:
$$\lim_{A \to \infty} I_{rect}(r) = 4L \cdot \frac{\pi}{2} = 2\pi L$$
Unlike a point source, which approaches infinite intensity at $r=0$, a large-area screen provides a **uniform, finite radiance** across the near-field.

#### **2. The Transition Boundary**
For screens where $A > 1$, the "Far-Field" (where $1/r^2$ becomes accurate) is pushed significantly further back. A common engineering heuristic is that the Far-Field begins at:
$$r_{limit} \approx 5 \times \sqrt{w^2 + h^2}$$
For an 85-inch TV ($w \approx 1.9\text{m}$), the Inverse Square Law is technically invalid until the observer is more than **10 meters** away.

#### **3. Architectural Impact**
In this regime, the display acts more like a **window** than a lamp. The light concentration ($I/A$) is low, but the total immersion is high. This explains why "Projector Screens" or "Large OLEDs" feel more comfortable at high brightness than a tiny, high-intensity LED bulb; the light is spatially distributed, reducing retinal peak-stress.


### **4. The Confinement Model: Evaluating Spatial Distribution ($L \cdot A \approx 1$)**

To isolate the geometric impact of screen size from raw power, we introduce the **Confinement Constraint**. We define the total luminous flux $\Phi$ as:

$$\Phi = L \cdot A = L \cdot (w \cdot h) \approx 1$$

Under this constraint, as the display area $A$ increases, the surface luminosity $L$ must decrease proportionally ($L = 1/A$). This allows us to compare a "Point-Like" source (high $L$, tiny $A$) to a "Surface-Like" source (low $L$, large $A$) with normalized energy output.

#### **Mathematical Implications**
Using the Confinement Model, our intensity formula $I_{rect}(r)$ becomes:

$$I_{rect}(r) = \frac{4}{w \cdot h} \cdot \arctan\left(\frac{w \cdot h}{4r \sqrt{r^2 + (w/2)^2 + (h/2)^2}}\right)$$

#### **Key Observations at $L \cdot A = 1$**

1. **The Point Source Limit ($A \to 0$):**
   As the area vanishes while maintaining $\Phi=1$, the model perfectly converges to the Inverse Square Law $1/r^2$. The light is "perfectly confined."

2. **The Near-Field Plateau ($A > 1$):**
   For large surfaces (e.g., $A = 4 \text{ m}^2$), the intensity near the screen ($r \to 0$) is capped at a finite value. Even though the total light emitted is the same as the point source, the **peak intensity** at the eye is significantly lower because the light is "diluted" over a massive solid angle.

3. **The NDI Sensitivity:**
   The **Near-Field Deviation Index (NDI)** reaches its maximum when $A > 1$. This confirms that for large-format displays, the $1/r^2$ model is not just slightly off—it is fundamentally the wrong tool for measuring human exposure.


```python
import numpy as np
from scipy.integrate import quad

class Display:
    """
    Model for a Rectangular Lambertian Emitter.
    Implements the Confinement Model where Total Flux (L * A) = 1.
    """
    def __init__(self, name, width, height):
        self.name = name
        self.w = width
        self.h = height
        self.area = width * height
        # Confinement Model: L * A = 1 => L = 1/A
        self.L = 1.0 / self.area 

    def intensity_rect(self, r):
        """Analytical solution for rectangular surface intensity."""
        term = (self.w * self.h) / (4 * r * np.sqrt(r**2 + (self.w/2)**2 + (self.h/2)**2))
        return 4 * self.L * np.arctan(term)

    def intensity_point(self, r):
        """Standard Inverse Square Law (Point Source) normalized to Total Flux = 1."""
        return 1.0 / (r**2)

    def calculate_ndi(self, r_min=0.2, r_max=2.0):
        """Calculates the Near-Field Deviation Index (Area between curves)."""
        area_diff, _ = quad(lambda r: np.abs(self.intensity_rect(r) - self.intensity_point(r)), r_min, r_max)
        return area_diff


    def calculate_avg_intensity(self, r):
        """
        Calculates the Average Intensity (Concentration) 
        across the physical surface area.
        """
        # I_avg = Total Intensity / Surface Area
        return self.intensity_rect(r) / self.area

# --- Comparison of Light Concentration ---
r_work = 0.5  # Standard desk distance (meters)





# --- Portfolio Comparison Execution ---
devices = [
    Display("Smartphone (6\")", 0.07, 0.15),
    Display("Laptop (14\")", 0.30, 0.20),
    Display("Monitor (27\")", 0.60, 0.35),
    Display("OLED TV (65\")", 1.45, 0.82),
    Display("LED Wall (120\")", 2.60, 1.50) # A > 1 case
]

print(f"{'Device':<20}|{'Area (m2)':<10}|{'NDI Value':<10}|{'I_eye':<10}|{'I_avg':<10}")
print("-" * 80)
for d in devices:
    ndi = d.calculate_ndi()
    i_eye = d.intensity_rect(r_work)
    i_avg = d.calculate_avg_intensity(r_work)
    print(f"{d.name:<20.4}|{d.area:<10.4f}|{ndi:<10.4f}|{i_eye:<10.4f}|{i_avg:<10.4f}")
```

```python
import matplotlib.pyplot as plt

# Distances from 0.15m to 2.5m for the plot
r_range = np.linspace(0.15, 2.5, 500)

plt.figure(figsize=(12, 7))

# 1. Plot the "Universal" Point Source (L*A = 1)
y_point = [d.intensity_point(r) for r in r_range]
plt.plot(r_range, y_point, label='Point Source Model ($1/r^2$)', 
         color='black', linestyle='--', linewidth=2, alpha=0.6)

# 2. Plot our Confinement Model Displays
colors = ['#FF9999', '#66B2FF', '#99FF99', '#FFCC99', '#FF66B2']
for i, d in enumerate(devices):
    y_rect = [d.intensity_rect(r) for r in r_range]
    plt.plot(r_range, y_rect, label=f'{d.name} (NDI: {d.calculate_ndi():.2f})', 
             color=colors[i], linewidth=2.5)

# Formatting for a "Professional Portfolio" look
plt.title('The Near-Field Intensity Bulge (Normalized Flux $L \\cdot A = 1$)', fontsize=14)
plt.xlabel('Radial Distance from Screen $r$ (meters)', fontsize=12)
plt.ylabel('Relative Intensity $I(r)$', fontsize=12)
plt.yscale('log') # Log scale is essential to see the near-field plateau vs far-field convergence
plt.ylim(0.1, 50)
plt.xlim(0.15, 2.5)
plt.grid(True, which="both", ls="-", alpha=0.2)
plt.legend(frameon=True, shadow=True)

# Annotations to explain the "Physics Shift"
plt.annotate('Far-Field: Models Converge', xy=(2.0, 0.25), xytext=(1.2, 0.5),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=5))
plt.annotate('Near-Field: Geometry Dilutes Intensity', xy=(0.2, 10), xytext=(0.4, 25),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=5))

plt.tight_layout()
plt.show()
```

```python
def calculate_far_field_boundary(display, error_threshold=0.05):
    """
    Finds the distance 'r' where the % difference between 
    I_rect and I_point is less than the threshold.
    """
    r_test = 0.1
    while r_test < 20.0: # Search up to 20 meters
        i_r = display.intensity_rect(r_test)
        i_p = display.intensity_point(r_test)
        
        # Calculate relative error
        error = np.abs(i_r - i_p) / i_p
        
        if error < error_threshold:
            return r_test
        r_test += 0.01
    return None

# --- Generate Usage Recommendations ---
print(f"{'Device':<20} | {'Far-Field Boundary (m)':<25} | {'Safe to use 1/r^2?'}")
print("-" * 75)
for d in devices:
    ff_boundary = calculate_far_field_boundary(d)
    status = "At living room dist." if ff_boundary < 3.0 else "Only in large halls"
    print(f"{d.name:<20} | {ff_boundary:<25.2f} | {status}")
```

This defines the **"Concentration Gap"**: Large screens are safer and more comfortable not because they are "dimmer" in total energy, but because their **geometric footprint** dilutes that energy across a massive solid angle, preventing high-density retinal "hotspots."


```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

def plot_3d_comparison(dev1, dev2, r_fixed=0.5):
    fig = plt.figure(figsize=(16, 7))
    
    # Grid setup (Observer moving +/- 0.5m around center)
    x = np.linspace(-0.5, 0.5, 80)
    y = np.linspace(-0.5, 0.5, 80)
    X, Y = np.meshgrid(x, y)

    for i, dev in enumerate([dev1, dev2]):
        ax = fig.add_subplot(1, 2, i+1, projection='3d')
        
        # Re-using our derived intensity function for off-axis points
        def f(x_v, y_v, z_v):
            return (x_v / np.sqrt(x_v**2 + z_v**2)) * np.arctan(y_v / np.sqrt(x_v**2 + z_v**2))
        
        # Compute Intensity at each (X, Y) point
        x1, x2 = -dev.w/2 - X, dev.w/2 - X
        y1, y2 = -dev.h/2 - Y, dev.h/2 - Y
        Z = 4 * dev.L * (f(x2, y2, r_fixed) - f(x1, y2, r_fixed) - f(x2, y1, r_fixed) + f(x1, y1, r_fixed))
        
        # Surface Plot
        surf = ax.plot_surface(X, Y, Z, cmap='magma', edgecolor='none', antialiased=True)
        
        ax.set_title(f'{dev.name}\n(Intensity Concentration $I_{{avg}} \\approx {dev.calculate_avg_intensity(r_fixed):.1f}$)', fontsize=12)
        ax.set_zlim(0, 10) # Locked Z-axis to show the massive difference in height
        ax.set_xlabel('X (m)')
        ax.set_ylabel('Y (m)')
        ax.set_zlabel('Intensity')

    plt.suptitle(f'Spatial Light Concentration at r = {r_fixed}m (Normalized Flux $LA=1$)', fontsize=16)
    plt.tight_layout()
    plt.show()

# Run the comparison between the smallest and largest devices
plot_3d_comparison(devices[0], devices[-1]) 
```
